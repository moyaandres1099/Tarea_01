### Por otra parte, este gráfico representa la distribución de la densidad de población por distrito del cantón de Puriscal, según el Censo 2022. Santiago presenta la mayor densidad, con 353,1 habitantes por km², seguido de San Antonio, con 335,7. En contraste, Chires registra la menor densidad, con 13,2 habitantes por km², seguido de Mercedes Sur, con 38,8. Estos resultados evidencian diferencias importantes en la distribución espacial de la población, los distritos de Santiago y San Antonio concentran más habitantes en relación con su superficie, mientras que Chires presenta una ocupación poblacional mucho más dispersa.

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt

AZUL = "#1F77B4"
NARANJA = "#FF7F0E"
VERDE = "Green"
GRIS = "lightgray"

distritos  = pd.read_csv("https://raw.githubusercontent.com/gf0657-programacionsig/2026-ii/refs/heads/main/datos/inec/distritos-2022.csv")

# Fila de Chires
chires = distritos[distritos["distrito"].str.strip() == "Chires"].iloc[0]

# Cada barra: lista de (nombre, valor, color, color del texto)
barras = {
    f"Población\n{chires['poblacion']:,.0f} habitantes": [
        ("Hombres", chires["hombres"], AZUL, "white"),
        ("Mujeres", chires["mujeres"], NARANJA, "white"),
    ],
    f"Viviendas\n{chires['viviendas']:,.0f} viviendas": [
        ("Ocupadas", chires["viviendas_ocupadas"], VERDE, "white"),
        ("Desocupadas", chires["viviendas_desocupadas"], GRIS, "black"),
    ],
}

fig, ax = plt.subplots(figsize=(10, 4.5))

for i, (etiqueta, partes) in enumerate(barras.items()):
    total = sum(valor for _, valor, _, _ in partes)
    inicio = 0
    for nombre, valor, color, color_texto in partes:
        pct = valor / total * 100
        ax.barh(i, pct, left=inicio, color=color, edgecolor="white", height=0.55)
        ax.text(inicio + pct / 2, i, f"{nombre}\n{valor:,.0f} ({pct:.1f}%)",
                ha="center", va="center", fontsize=10, color=color_texto)
        inicio += pct

ax.set_yticks(range(len(barras)))
ax.set_yticklabels(list(barras.keys()), fontsize=11)
ax.invert_yaxis()   # Población arriba, Viviendas abajo
ax.set_xlim(0, 100)
ax.set_xlabel("Porcentaje del total (%)")
ax.spines[["top", "right"]].set_visible(False)

# Tasa de crecimiento como recuadro dentro del gráfico
ax.text(0.5, -0.30, f"Tasa de crecimiento de la población: {chires['tasa_crecimiento_poblacion']:.2f} %",
        transform=ax.transAxes, ha="center", va="center", fontsize=12, fontweight="bold",
        color=VERDE, bbox={"boxstyle": "round,pad=0.5", "facecolor": "white", "edgecolor": VERDE})

ax.set_title("Caracterización de la población del distrito de Chires, Censo 2022")

fig.tight_layout()
plt.show()
```

### El gráfico muestra que el distrito de Chires tiene una población de 3.022 habitantes, con una distribución entre hombres (51,3 %) y mujeres (48,7 %). En cuanto a las viviendas, de las 1.329 registradas, el 81,9 % están ocupadas, mientras que el 18,1 % se encuentran desocupadas. Además, la tasa de crecimiento poblacional es de -0,03 %, lo que indica una ligera disminución de la población.

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt

AZUL = "#1F77B4"
NARANJA = "#FF7F0E"
VERDE = "Green"
GRIS = "lightgray"

distrito = "Mercedes Sur"

distritos  = pd.read_csv("https://raw.githubusercontent.com/gf0657-programacionsig/2026-ii/refs/heads/main/datos/inec/distritos-2022.csv")

# Fila del distrito
fila = distritos[distritos["distrito"].str.strip() == distrito].iloc[0]

# Cada barra: lista de (nombre, valor, color, color del texto)
barras = {
    f"Población\n{fila['poblacion']:,.0f} habitantes": [
        ("Hombres", fila["hombres"], AZUL, "white"),
        ("Mujeres", fila["mujeres"], NARANJA, "white"),
    ],
    f"Viviendas\n{fila['viviendas']:,.0f} viviendas": [
        ("Ocupadas", fila["viviendas_ocupadas"], VERDE, "white"),
        ("Desocupadas", fila["viviendas_desocupadas"], GRIS, "black"),
    ],
}

fig, ax = plt.subplots(figsize=(10, 4.5))

for i, (etiqueta, partes) in enumerate(barras.items()):
    total = sum(valor for _, valor, _, _ in partes)
    inicio = 0
    for nombre, valor, color, color_texto in partes:
        pct = valor / total * 100
        ax.barh(i, pct, left=inicio, color=color, edgecolor="white", height=0.55)
        ax.text(inicio + pct / 2, i, f"{nombre}\n{valor:,.0f} ({pct:.1f}%)",
                ha="center", va="center", fontsize=10, color=color_texto)
        inicio += pct

ax.set_yticks(range(len(barras)))
ax.set_yticklabels(list(barras.keys()), fontsize=11)
ax.invert_yaxis()   # Población arriba, Viviendas abajo
ax.set_xlim(0, 100)
ax.set_xlabel("Porcentaje del total (%)")
ax.spines[["top", "right"]].set_visible(False)

# Tasa de crecimiento como recuadro dentro del gráfico
ax.text(0.5, -0.30, f"Tasa de crecimiento de la población: {fila['tasa_crecimiento_poblacion']:.2f} %",
        transform=ax.transAxes, ha="center", va="center", fontsize=12, fontweight="bold",
        color=VERDE, bbox={"boxstyle": "round,pad=0.5", "facecolor": "white", "edgecolor": VERDE})

ax.set_title(f"Caracterización de la población del distrito de {distrito}, Censo 2022")

fig.tight_layout()
plt.show()
```

### El gráfico caracteriza la población del distrito de Mercedes Sur con los datos del Censo 2022, en dos barras que suman 100 % cada una:

### Población por sexo: el distrito tiene 7 056 habitantes, de los cuales 3 455 son hombres (49.0 %) y 3 601 son mujeres (51.0 %). Hay una ligera mayoría de mujeres, aunque la distribución es casi equilibrada.

### Estado de las viviendas: de las 2 896 viviendas, 2 539 (87.7 %) están ocupadas y 357 (12.3 %) están desocupadas. Casi 9 de cada 10 viviendas son habitadas, y las desocupadas son una minoría, aunque no despreciable.

### Tasa de crecimiento: la población crece a un ritmo de 1.68 %, lo que indica un aumento sostenido de habitantes
